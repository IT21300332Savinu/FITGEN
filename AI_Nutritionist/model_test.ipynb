Full Model Test

```python
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load the model
model = joblib.load("calorie_prediction_model.pkl")
meal_embeddings = np.load("meal_embeddings.npy")
df = pd.read_csv("meal_suggestion_meal_plans.csv")
meal_suggestion_model = SentenceTransformer("meal_suggestion_sentence_model")

def get_meal_plan_for_calorie(calorie_target):
    query = f"A healthy meal plan for a person needing {calorie_target} calories"
    query_embedding = meal_suggestion_model.encode([query])
    similarities = cosine_similarity(query_embedding, meal_embeddings)[0]
    best_index = np.argmax(similarities)
    return df.iloc[best_index][['Breakfast Suggestion', 'Lunch Suggestion', 'Dinner Suggestion', 'Snack Suggestion']]
```

```python
sample = pd.DataFrame([{
    'Age': 30,
    'Gender': 'Male',
    'Height': 175,
    'Weight': 70,
    'Activity Level': 'Moderately active',
    'Dietary Preference': 'Veg',
    'Budget Preferences': 'Medium',
    'Acne': 0,
    'Diabetes': 0,
    'Heart Disease': 0,
    'Hypertension': 0,
    'Kidney Disease': 0,
    'Weight Gain': 0,
    'Weight Loss': 1
}])
```

```python
# Predict
predicted_calories = model.predict(sample)
print(f"Predicted daily calorie need: {predicted_calories[0]:.2f} kcal")
```

```python
suggested_meals = get_meal_plan_for_calorie(predicted_calories)
print("Suggested Meal Plan:")
print(suggested_meals)
```
